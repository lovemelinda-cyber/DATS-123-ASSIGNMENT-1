```python
#QUESTION 3A)
arr = [21,10,45,32,67,5,18]
def linear_search(arr,target):
    for i in range(len(arr)):
       if arr[i] == target:
           return i 
    return -1
result=linear_search (arr,32)
print(f"a)Linear Search:Value 32 found at index {result}")
```

```python
#QUESTION 3B)
def binary_search(arr,target):
    left=0
    right=len(arr) - 1
   
    while left<= right:
        mid = (left + right )// 2

        if arr[mid] == target:
            return mid 
        elif arr [mid]< target:
            left=mid + 1 
        else: 
            right = mid -1
    
    return -1 
sorted_arr = sorted(arr)
result= binary_search(sorted_arr,45)
print(f"\nb) Binary Search: Sorted array = {sorted_arr}")
print(f" Value 45 found at index {result}")
```

```python
# QUESTION 3C)
def bubble_sort(arr):
    n=len(arr)

    for i in range(n):
        for j in range (0,n-i-1):
            if arr[j]> arr[j+1]:
                arr[j],arr[j+1]= arr[j+1],arr[j]
   
    return arr 
original_arr =[21,10,45,32,67,5,18]
sorted_result = bubble_sort(original_arr.copy())
print(f"\nc) Bubble Sort:")
print(f" Sorted:{sorted_result}")
```

```python
#QUESTION FIVE 
#a) the most appropriate data structure would be a queue as it follows the fifo (first in first out) prinicple which 
# which will be perfect for managing students waiting in like a finance office.
#b)
class StudentQueue:
   
    def _init_(self):
        self.queue=[]

    def add_student(self,student_name ):
        self.queue.append(student_name)
        print(f"Student'{student_name}' added to the queue")

    def serve_student(self):
        if self.is_empty():
           print("Queue is empty! No students to serve.")
           return None

        served_student=self.queue.pop(0)
        print(f"serving student: '{served_student}'")
        return served_student
        
    def display_queue(self):
        if self.is_empty():
            print("queue us empty!")
        else:
            print(f"Current queue({len(self.queue)}student(s)):")
            for i,student in enumerate(self.queue,1):
                print(f"{i}.{student }")

    def is_empty(self):
        return len(self.queue)==0
```
